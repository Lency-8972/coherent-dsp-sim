```python
import numpy as np
import matplotlib.pyplot as plt

# 生成一条 50 Hz 正弦波
fs = 1000                       # 采样率 1000 Hz
t = np.arange(0, 1, 1/fs)       # 1 秒的时间轴
y = np.sin(2 * np.pi * 50 * t)

# 画图
plt.plot(t[:100], y[:100])
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('W0 check: 50 Hz sine wave')
plt.grid(True)
plt.show()
print('numpy version:', np.__version__)
```
